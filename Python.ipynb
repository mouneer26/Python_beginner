## Simple Calculator using Functions

This guide will walk you through creating a basic calculator that performs addition, subtraction, multiplication, and division using Python functions.

First, let's define the functions for each arithmetic operation: addition, subtraction, multiplication, and division.

```python
def add(x, y):
  """Adds two numbers"""
  return x + y

def subtract(x, y):
  """Subtracts two numbers"""
  return x - y

def multiply(x, y):
  """Multiplies two numbers"""
  return x * y

def divide(x, y):
  """Divides two numbers. Handles division by zero."""
  if y == 0:
    return "Error: Cannot divide by zero!"
  return x / y
```

Now, let's allow the user to choose an operation and input two numbers. We'll then use our defined functions to perform the calculation.

```python
print("Select operation:")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")

while True:
  choice = input("Enter choice(1/2/3/4): ")

  if choice in ('1', '2', '3', '4'):
    try:
      num1 = float(input("Enter first number: "))
      num2 = float(input("Enter second number: "))
    except ValueError:
      print("Invalid input. Please enter numbers only!")
      continue

    if choice == '1':
      print(f"{num1} + {num2} = {add(num1, num2)}")
    elif choice == '2':
      print(f"{num1} - {num2} = {subtract(num1, num2)}")
    elif choice == '3':
      print(f"{num1} * {num2} = {multiply(num1, num2)}")
    elif choice == '4':
      result = divide(num1, num2)
      print(f"{num1} / {num2} = {result}")
    break
  else:
    print("Invalid Input. Please enter a valid choice (1/2/3/4).")
```

## Multiplication Table Generator

This section generates a multiplication table for a number up to a specified range.

```python
# Get input from the user for the number and the range
number_to_multiply = int(input("Enter the number for the multiplication table: "))
range_limit = int(input("Enter the range limit (e.g., 10 for 1 to 10): "))
```

```python
# Generate and print the multiplication table
print(f"\nMultiplication Table for {number_to_multiply} up to {range_limit}:")
for i in range(1, range_limit + 1):
    product = number_to_multiply * i
    print(f"{number_to_multiply} x {i} = {product}")
```

```python
number_to_add = int(input("Enter the number for the addition table:"))
range_limit = int(input("Enter the range limit (e.g., 10 for 1 to 10): "))

print(f"\nAdditiontion Table for {number_to_add} up to {range_limit}:")
for i in range(1, range_limit + 1):
    product = number_to_add + i
    print(f"{number_to_add} + {i} = {product}")
```

## Pie Chart Generation

This section demonstrates how to create a pie chart using `matplotlib.pyplot`.

A pie chart is a circular statistical graphic, divided into slices to illustrate numerical proportion. In a pie chart, the arc length of each slice (and consequently its central angle and area) is proportional to the quantity it represents. Before plotting, let's define some sample data.

```python
import matplotlib.pyplot as plt

# Sample data for the pie chart
labels = ['Biology', 'Chemistry', 'Physics', 'Mathematics']
sizes = [15, 30, 45, 10] # Proportions for each category
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue'] # Colors for each slice
explode = (0, 0.5, 0, 0)  # 'explode' the 2nd slice (Category B) slightly
```

```python
# Create the pie chart
fig1, ax1 = plt.subplots()
ax1.pie(sizes, explode=explode, labels=labels, colors=colors,
        autopct='%1.1f%%', shadow=True, startangle=90)
ax1.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.

plt.title("Distribution of Courses")
plt.show()
```
